```python
import sqlite3
```

```python
import pandas as pd

df = pd.read_csv('hasil_segmentasi_nasabah.csv')
display(df.head())
```

```python
koneksi = sqlite3.connect('database_nasabah.db')
df.to_sql('tabel_nasabah', koneksi, if_exists='replace', index=False)
```

```python
koneksi = sqlite3.connect('database_nasabah.db')
query = "SELECT * FROM tabel_nasabah"
df_1 = pd.read_sql(query, koneksi)
koneksi.close()

print("--- Ringkasan Angka (Numerik) ---")
display(df_1.describe())

print("\n--- Ringkasan Kategori (Teks) ---")
display(df_1.describe(include=['object']))
```
